```python
# PROJECT 1

import tkinter as tk
from tkinter import ttk
from tkinter import messagebox as msgbox
import random

class Employee:
    def __init__(self):
        self.employees = ["Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali", "Andrew Ugwu"
             , "Stella Mankinde", "Jane Akibo", "Ago James", "Michell Taiwo", "Abraham Jones"
             , "Nicole Anide", "Kosi Korso", "Adele Martins", "Emmanuel Ojo", "Ajayi Fatima"]

        self.tasks = ["Loading", "Transporting", "Reveiwing Orders", "Customer Service", "Delivering Items"]

    def check_employee(self, name):
        if name.title() in self.employees:
            attendance_msg = self.take_attendance(name)
            task_msg = self.assign_task()
            return f"{attendance_msg}\n{task_msg}"
        else:
            return self.refuse_access()

    def take_attendance(self, name):
        return f"Attendance for {name} has been recorded."
    
    def assign_task(self):
        assignment = random.choice(self.tasks)
        return f"Your task is {assignment}"

    def refuse_access(self):
        return f"Your name doesn't match records. Access has been denied."

system = Employee()
def handle_login():
    user_name = name_entry.get()
    result = system.check_employee(user_name)
    msgbox.showinfo("Attendance Verification", result)
    
root = tk.Tk()
root.title("Jane Logistics Staff Attendance")
root.geometry("500x300")

name_label = tk.Label(root, text="Enter Employee Full Name:")
name_label.pack()
name_entry = tk.Entry(root)
name_entry.pack()

login_button = tk.Button(root, text="Login", command=handle_login)
login_button.pack()

root.mainloop()
```
