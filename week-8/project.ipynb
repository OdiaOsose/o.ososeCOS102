```python
import tkinter as tk
from tkinter import ttk

# Backend
class Zenith:
    def mutual_services(self):
        return ["Lines of credit", "Investment management and accounts", "Insurance"]
    
    def unique_services(self):
        return []

class Retail(Zenith):
    def unique_services(self):
        return ["Retirement and education accounts", "Loans and mortgages", "Checking and saving"]

class Commercial(Zenith):
    def unique_services(self):
        return ["Advisory services"]

class Global(Zenith):
    def mutual_services(self):
        return ["None (Independent Division)"]
    def unique_services(self):
        return ["Multi-currency management services and products", "Foreign currency accounts", "Foreign currency credit cards", "Transborder advisory services", "Liquidity management"]


# Bridge
def fetch_services():
    name = name_entry.get()
    division = division_box.get()
    
    # Polymorphic object creation
    if division == "Retail Banking":
        obj = Retail()
    elif division == "Commercial Banking":
        obj = Commercial()
    else:
        obj = Global()
        
    # Clear old text and insert new results
    output_text.delete("1.0", tk.END)
    output_text.insert(tk.END, f"Employee: {name}\nDivision: {division}\n\n")
    
    output_text.insert(tk.END, "MUTUAL SERVICES:\n")
    for s in obj.mutual_services():
        output_text.insert(tk.END, f"- {s}\n")
        
    output_text.insert(tk.END, "\nUNIQUE SERVICES:\n")
    for s in obj.unique_services():
        output_text.insert(tk.END, f"- {s}\n")


# Frontend
root = tk.Tk()
root.title("Zenith Bank")

# Inputs
tk.Label(root, text="Employee Name:").pack(pady=2)
name_entry = tk.Entry(root, width=30)
name_entry.pack(pady=2)

tk.Label(root, text="Division:").pack(pady=2)
division_box = ttk.Combobox(root, values=["Retail Banking", "Global Banking", "Commercial Banking"], state="readonly")
division_box.current(0)
division_box.pack(pady=2)

# Action Button
tk.Button(root, text="Submit", command=fetch_services).pack(pady=10)

# Output Display
output_text = tk.Text(root, width=50, height=12)
output_text.pack(pady=5)

root.mainloop()
```
