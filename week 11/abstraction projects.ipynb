```python
from abc import ABC, abstractmethod

class coup_de_ecriva(ABC):
    
    @abstractmethod
    def Fan_Page(self):
        pass


class FC_Cirok(coup_de_ecriva):
    def Fan_Page(self):
        print("Welcome to FC Cirok")

class Madiba_FC(coup_de_ecriva):
    def Fan_Page(self):
        print("Welcome to Madiba FC")

class Blue_Jay_FC(coup_de_ecriva):
    def Fan_Page(self):
        print("Welcome to Blue Jay FC")

class TSG_Walker(coup_de_ecriva):
    def Fan_Page(self):
        print("Welcome to TSG Walker")

def main():
    print("Coup De Ecriva Football Portal")
    print("Available Clubs: FC Cirok, Madiba FC, Blue Jay FC, TSG Walker\n")
    
    user_choice = input("Enter the name of the club you support: ").strip().lower()

    # Match input to the correct subclass object
    if user_choice in ["fc cirok", "fc_cirok", "cirok"]:
        club_object = FC_Cirok()
    elif user_choice in ["madiba fc", "madiba_fc", "madiba"]:
        club_object = Madiba_FC()
    elif user_choice in ["blue jay fc", "blue_jay_fc", "blue jay"]:
        club_object = Blue_Jay_FC()
    elif user_choice in ["tsg walker", "tsg_walker", "tsg", "walker"]:
        club_object = TSG_Walker()
    else:
        print("Error: Club not recognized. Please check your spelling and try again.")
        return

    # Invoke the Fan_Page method using polymorphism
    club_object.Fan_Page()

main()
```

```python
import tkinter as tk
from tkinter import ttk
from abc import ABC, abstractmethod

class External_Vendors(ABC):
    @abstractmethod
    def menu(self):
        pass

class Cooperative_Hostel(External_Vendors):
    def menu(self):
        return [
            ("Jollof Rice and Stew", 200),
            ("White Rice and Stew", 200),
            ("Fried Rice", 200),
            ("Salad", 100),
            ("Plantain", 100)
        ]

class Faith_Hostel(External_Vendors):
    def menu(self):
        return [
            ("Fried Rice", 400),
            ("White Rice and Stew", 400),
            ("Jollof Rice", 400),
            ("Beans", 200),
            ("Chicken", 1000)
        ]

class Student_Center(External_Vendors):
    def menu(self):
        return [
            ("Chicken Fried Rice", 800),
            ("Pomo Sauce", 300),
            ("Spaghetti Jollof", 500),
            ("Amala/Ewedu", 500),
            ("Semo with Eforiro Soup", 500)
        ]

def display_vendor_menu():
    selected_vendor = vendor_box.get()
    
    # Polymorphic object creation based on GUI choice
    if selected_vendor == "Cooperative Cafeteria":
        vendor_object = Cooperative_Hostel()
    elif selected_vendor == "Faith Hostel Cafeteria":
        vendor_object = Faith_Hostel()
    else:
        vendor_object = Student_Center()
        
    # Extract data using the common menu method
    menu_items = vendor_object.menu()
    
    # Refresh the text box output area
    output_text.delete("1.0", tk.END)
    output_text.insert(tk.END, f"MENU FOR: {selected_vendor.upper()}\n")
    output_text.insert(tk.END, f"{'Main Meal':<30}{'Price (N)':<15}\n")
    output_text.insert(tk.END, "-" * 45 + "\n")
    
    for meal, price in menu_items:
        output_text.insert(tk.END, f"{meal:<30}N{price:<15}\n")

root = tk.Tk()
root.title("PAU External Food Vendors")

tk.Label(root, text="Select Vendor of Interest:", font=("Arial", 11)).pack(pady=5)
vendor_box = ttk.Combobox(root, values=["Cooperative Cafeteria", "Faith Hostel Cafeteria", "Student Centre Cafeteria"], state="readonly", width=28, font=("Arial", 10))
vendor_box.current(0)
vendor_box.pack(pady=5)

tk.Button(root, text="Display Menu", font=("Arial", 10, "bold"), command=display_vendor_menu).pack(pady=10)

output_text = tk.Text(root, width=48, height=12, font=("Courier", 10))
output_text.pack(pady=10, padx=10)

root.mainloop()
```
