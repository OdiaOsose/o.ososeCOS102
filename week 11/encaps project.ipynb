```python
import pandas as pd

class StudentRequest:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade
        
        self.__error_message = None 
        
        # Validation rules
        if self.age > 25:
            self.__error_message = f"Student: {self.name} (Age: {self.age}) exceeds the permissible age range (> 25)."

    # Getter method to read the encapsulated error message
    def get_error_message(self):
        return self.__error_message

class StudentInformationSystem:
    def __init__(self, file_path="SIS.csv"):
        self.file_path = file_path
        self.pirates_bucket = []
        self.yankees_bucket = []
        self.bulls_bucket = []
        self.age_violations = []

    def classify_students(self):
        try:
            # Read the input CSV
            df = pd.read_csv(self.file_path)
        except FileNotFoundError:
            print(f"File Error: '{self.file_path}' was not found. Please place it in the same directory.")
            return

        # Loop through each student row
        for index, row in df.iterrows():
            # get standard columns from the CSV file
            name = row['Name']
            age = int(row['Age'])
            grade = row['Grade']
            
            # Create an object instance for the student record
            request = StudentRequest(name, age, grade)
            
            # for encapsulated age exceptions via the Getter Method
            if request.get_error_message():
                self.age_violations.append(request.get_error_message())
                
            # Run age classification ranges
            elif 14 < age < 18:
                self.pirates_bucket.append(row)
            elif 18 <= age < 22:
                self.yankees_bucket.append(row)
            elif 22 <= age <= 25:
                self.bulls_bucket.append(row)

        # Step 5: Process output files and show on screen
        self.save_and_display_category("The_Pirates.csv", self.pirates_bucket)
        self.save_and_display_category("The_Yankees.csv", self.yankees_bucket)
        self.save_and_display_category("The_Bulls.csv", self.bulls_bucket)

        # Print all caught encapsulation errors
        if self.age_violations:
            print("ENCAPSULATED AGE ERRORS RETURNED VIA GETTER METHOD:")
            for error in self.age_violations:
                print(f" {error}")

    def save_and_display_category(self, filename, record_list):
        print(f"\nDISPLAYING FILE CONTENTS: {filename}")
        
        if record_list:
            # Convert filtered list of rows back into a DataFrame
            category_df = pd.DataFrame(record_list)
            # Save to a fresh CSV file
            category_df.to_csv(filename, index=False)
            # Display it
            print(category_df.to_string(index=False))
        else:
            print("No students found matching this age range.")

pau_sis = StudentInformationSystem("SIS.csv")
    
# Run the classification program
pau_sis.classify_students()
```
